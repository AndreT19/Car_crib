```python
import pandas as pd
import matplotlib.pyplot as plt

"""Mostrando a quantidade de carros por ano de fabricação"""

car_data = pd.read_csv('../vehicles_us.csv')  # lendo os dados
plt.hist(car_data['model_year'], bins=65, edgecolor='black')
plt.xlabel('Model year')
plt.ylabel('Quantity of Cars')
plt.title('Distribution of Cars by Year of Manufacture')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

"""Mostrando a relação entre o ano de fabricação e o preço do carro"""

car_data = pd.read_csv('../vehicles_us.csv')  # lendo os dados


plt.scatter(car_data['model_year'], car_data['price'], alpha=0.5)
plt.xlabel('Year of Manufacture')
plt.ylabel('Price')
plt.title('Relationship between Year of Manufacture and Price')
plt.show()
```
